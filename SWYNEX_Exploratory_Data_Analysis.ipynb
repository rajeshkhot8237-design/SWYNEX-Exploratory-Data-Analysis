```python
import pandas as pd

df = pd.read_csv('iris_cleaned (1).csv')

print("Dataset Shape:", df.shape)
print("\nFirst 5 Rows:")
display(df.head())

print("\nStatistical Summary:")
display(df.describe())

print("\nMissing Values:")
print(df.isnull().sum())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Numeric columns only
numeric_df = df.select_dtypes(include='number')

# Chart 1: Feature distributions
numeric_df.hist(figsize=(10, 7), edgecolor='black')
plt.suptitle("Iris Dataset - Feature Distributions")
plt.tight_layout()
plt.show()

# Chart 2: Correlation heatmap
plt.figure(figsize=(8, 5))
sns.heatmap(numeric_df.corr(), annot=True, cmap="coolwarm")
plt.title("Feature Correlation")
plt.show()

# Chart 3: Box plots
numeric_df = df.select_dtypes(include='number')

numeric_df.plot(
    kind='box',
    subplots=True,
    layout=(2, 3),
    figsize=(12, 7),
    sharex=False,
    sharey=False
)

plt.tight_layout()
plt.show()
```

```python
# 5 Insights from Iris Dataset

print("EDA INSIGHTS")
print("1. Dataset contains", df.shape[0], "rows and", df.shape[1], "columns.")

numeric_df = df.select_dtypes(include='number')

print("\n2. Average values of numeric features:")
print(numeric_df.mean().round(2))

print("\n3. Minimum values:")
print(numeric_df.min())

print("\n4. Maximum values:")
print(numeric_df.max())

print("\n5. Strongest positive correlation:")
corr = numeric_df.corr()
pairs = corr.where(
    __import__('numpy').triu(
        __import__('numpy').ones(corr.shape), k=1
    ).astype(bool)
).stack()

if not pairs.empty:
    print(pairs.idxmax(), ":", round(pairs.max(), 2))
else:
    print("Not enough numeric columns to calculate correlation.")
```
